## Bayesian Statistic

* A statistical approach that uses **Bayes' theorem** to update probabilities on new data, incoparating prior knowledge and uncertainty.
* The probability of observing data given a specific hypothesis or parameter value


Problem:
* A rare disease affects 1 in 10,000 people. A diagnostic test for the disease is 99% sensitive (correctly identifies those with the disease) and 98% specific (correctly identifies those without the disease).
A person takes the test and receives a positive result. What is the probability that they actually have the disease?

**(Hint: Use Bayes' theorem and consider the probability of testing positive for both diseased and non-diseased individuals.)**


P(A/B) = P(B/A) * P(A) / P(B)

* P(A/B):probability the person has the disease given a positive result
* P(B/A):Probability of testing positive given that the person has the disease(sensitivity)
* P(A):is the prior probability of having the disease
* P(B): the total probability of testing positive
