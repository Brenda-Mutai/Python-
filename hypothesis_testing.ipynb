### t-testing
> we do random testing to avoid biasness
* 

```python
import numpy as np #arithmetic calculations
from scipy import stats # to perfom hypothesis testing
```

```python
sample= [122.09, 100.64, 125.77, 120.32, 118.25,  
        96.47, 111.4 ,  80.66, 110.77, 111.14, 
        102.9, 114.54,  88.09,  98.59,  87.07, 
       110.43, 101.9 , 123.89,  97.03, 116.23, 
        108.3, 112.82, 119.57, 131.38, 128.39]
```

```python
sample_mean=np.mean(sample)
sample_mean
```

```python
population_mean=100
population_mean
```

```python
alternative_hypothesis = 'there is a difference between the means and therefore we can conclude that the training worked'
null_hypothesis = 'there is no difference between the means and therefore we can conclude that the training did not work'
```

> did the training work?
yes, based on what we have the training sample mean and population is a slight difference
,alternative hypothesis should be done

*  **alternative hypothesis** there is a difference between the means and therefore we can conclude that the training worked
*  **null hypothesis** there is no difference between the means and therefore we can conclude that the training did not work

```python
alpha=0.05 #significance level
```

we now calculate the t statistic

t=(x_bar-mu)/(sample_sd/np.sqrt(n))

```python
sample_size=len(sample)#no of observations
sample_size
```

```python
sample_sd=np.std(sample, ddof=1)
sample_sd
```

```python
t_statistic=(sample_mean-population_mean)/(sample_sd/np.sqrt(sample_size))
t_statistic
```

```python
p_value=stats.t.sf(t_statistic, df=sample_size-1)
p_value
```

```python
if p_value<alpha:
    print(f'Based on the current evidence, we reject the null hypothesis and conclude that {alternative_hypothesis}')
else:
    print(f'Based on the current evidence, we fail to reject the null hypothesis and coclude that {null_hypothesis}')
```

```python
t_results=stats.ttest_1samp(a=sample, popmean=population_mean)
t_results
```

```python
control_variance=control.var()
control_variance
```

```python
experimental_variance=experimental.var()
experimental_variance
```

```python
control_df=len(control)-1
control_df
```

```python
experimental_df=len(experimental)-1
experimental_df
```

```python
pooled_var=((control_df*control_variance)+(experimental_df*experimental_variance))/(control_df+experimental_df)
pooled_var
```

```python
negative_t_statistic=(experimental_mean-control_mean)/(np.sqrt(pooled_var*((1/len(experimental))+(1/len(control)))))
negative_t_statistic
```

```python
positive_t_statistic=(control_mean-experimental_mean)/(np.sqrt(pooled_var*((1/len(experimental))+(1/len(control)))))
positive_t_statistic
```
