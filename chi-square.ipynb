**Chi-Square test**
> **A statistical test used to determine if there's a significant  association between two categorical variables.**

>  This test is used for making claims about the frequencies of categorical data.
* Because it is testing frequencies rather than population parameters, this test is known as a non-parametric test.
* **p-value <0.05** :Reject the null hypothesis(significant association exists)
* **p-value >0.05** :Fail to reject the null hypothesis(no significant association exists)


```python
# importing libraries
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
# Data from superbowl
data = pd.read_csv(r"superbowl.csv")
data.head()
```

```python
data['Coin Toss Winner'].value_counts()
```

```python
data['Game Winner'].value_counts()
```

```python
coin_toss_counts=data['Coin Toss Outcome'].value_counts().sort_index()
coin_toss_counts
```

```python
fair_coin_expected=[sum(coin_toss_counts)/2,sum(coin_toss_counts)/2]
fair_coin_expected
```

```python
fair_coin_observed=coin_toss_counts.values
fair_coin_observed
```

```python
#testing whether its a fair coin
fair_coin_result=stats.chisquare(fair_coin_observed,fair_coin_expected)
fair_coin_result
```

```python
fair_gamewinner_counts=data['Game Winner'].value_counts()
fair_gamewinner_counts
```

```python
observed = fair_gamewinner_counts.values
observed
```

```python
expected=[sum(fair_gamewinner_counts)/2,sum(fair_gamewinner_counts)/2]
expected
```

```python
results=stats.chisquare(observed,expected)
results
```
